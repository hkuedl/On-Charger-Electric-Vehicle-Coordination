#### Dependencies

```python
import glob
import pandas as pd
from datas import load_data, setup_seed
from charge import Charger
from interface import Interface
from microgrid import Microgrid
from run import System
```

#### Data Import and System Setup

```python
seed = 1
period = 48
setup_seed(seed)
postfix = '*.csv'
ev_path = 'dataset/ev_load/'
lmp_price_upper = pd.read_csv('dataset/lmp_price.csv').copy()
lmp_price_lower = pd.read_csv('dataset/lmp_price.csv').copy()
pv_generation = pd.read_csv('dataset/pv_generation.csv').copy()
base_load = pd.read_csv('dataset/base_load.csv').copy()

files = glob.glob(ev_path + postfix)
choose = [file.split('/')[-1].replace('.csv','') for file in files]
ev_datas = load_data(path=ev_path, postfix='*.csv', choose=choose)

chargers = []
for i in range(len(choose)):
    ev_load = ev_datas[i]
    charge = Charger(ev_data = ev_load,
                lmp_price_data = lmp_price_lower,
                period = period,
                path = choose[i])
    chargers.append(charge)
    
interface = Interface(chargers = chargers,
        period = period)

micro_grid = Microgrid(lmp_price_data = lmp_price_upper,
                base_data = base_load,
                pv_data = pv_generation,
                period = period)

operator = System(micro_grid = micro_grid,
                 chargers = interface)
```
